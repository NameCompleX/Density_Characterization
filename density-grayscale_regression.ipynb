```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

import matplotlib.pyplot as plt

# 读取文件
file_path = "F:/B208/Sample/pku2025/0418/gray-density.txt"
data = pd.read_csv(file_path, sep="\t", header=None, names=["x", "y"])

# 提取x和y
x = data["x"].values.reshape(-1, 1)
y = data["y"].values

# 创建线性回归模型并拟合数据
model = LinearRegression()
model.fit(x, y)

# 获取回归系数和截距
slope = model.coef_[0]
intercept = model.intercept_

# 打印结果
print(f"回归方程: y = {slope:.4f}x + {intercept:.4f}")
score = model.score(x, y)
print("模型得分:", score)
# 可视化
#plt.xlim(35,75)
#plt.ylim(3.0,8.0)
plt.scatter(x, y, color="blue", label="Real Density")
plt.plot(x, model.predict(x), color="red", label="Regression line")
plt.xlabel("Gray Scale")
plt.ylabel("Density")
plt.legend()
plt.title("Linear Regression")
plt.show()
```
